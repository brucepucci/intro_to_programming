## Example Exercise
name: add

arguments:
    * a (numeric),
    * b (numeric),
    * inverse (bool; optional: default=False)
    
returns: a + b or a + b * -1 if inverse

```python
def add(a, b, inverse=False):
    if inverse:
        return (a + b) * -1
    else:
        return a + b
```

```python
assert add(1, 2) == 3
assert add(1, 2, True) == -3
assert add(1, 2, inverse=True) == -3
```

## Exercise #1

name: calc_paycheck

arguments:
    * hours (int),
    * wage (float),
    * ot (bool; optional: default=False),
    
returns: calcuate the total weekly pay for a worker with the formula `hours` * `wage`. If the worker is overtime eligible (denoted by parameter `ot`) they will reveive 150% of their hourly wage for any time over 40 hours.

```python
def calc_paycheck(hours, wage, ot=False):
    if ot and hours > 40:
        return (40 * wage) + ((hours - 40) * (1.5 * wage))
    else:
        return hours * wage
```

```python
assert calc_paycheck(40, 10.) == 400
assert calc_paycheck(41, 10.) == 410
assert calc_paycheck(41, 10., ot=True) == 415
assert calc_paycheck(0, 10.) == 0
assert calc_paycheck(80, 20., ot=True) == 2000
```

aote: You MUST call the function `calc_paycheck` from exercise 1.

```python
def better_deal(lump_sum, hours, wage, ot=False):
    if lump_sum >= calc_paycheck(hours, wage, ot):
        return 'lump sum'
    else:
        return 'hourly wage'
```

## Exercise #3

name: calc_distance

arguments:
    * point_a (tuple of length 2, ie (4, 2), where index 0 is x-coordinant in a 2d coordinate plane)
    * point_b (same as point_a)
    
returns: The distance between the 2 points in a 2d coordinate plane

```python
def calc_distance(point_a, point_b):
    return ((point_a[0] - point_b[0])**2 + (point_a[1] - point_b[1])**2)**.5
```

```python
assert calc_distance((0, 0), (0, 1)) == 1
assert calc_distance((0, 0), (0, 0)) == 0
assert calc_distance((0, 0), (0, -1)) == 1
assert calc_distance((3, 2), (9, 7)) == 61**.5
```

## Exercise #4

name: closest_city

arguments:
    * taget_city (string, must be a key in `city_coordinates` which is defined below)
    
returns: The name of the nearest city to the target city. Use the function `calc_distance` that you defined in ecercise #3. Consider the dictionary `city_coordinates` given below a global variable and use it's values as the location of each city defined by it's respective key. You must run the cell below to register the dictionary city coordinates

```python
city_coordinates = {
    'Hong Kong': (9, 10),
    'New York': (2, 2),
    'Los Angeles': (-10, 12),
    'Tokyo': (0, 2),
    'London': (-3, -3),
    'Toronto': (5, 5),
    'Rome': (-14, 0),
    'Paris': (10, 10),
}
```

```python
def closest_city(target_city):
    distances = dict()
    for k, v in city_coordinates.items():
        if k != target_city:
            distances[k] = calc_distance(city_coordinates[target_city], v)
    return min(distances.keys(), key=(lambda k: distances[k]))
```

```python
assert closest_city('Tokyo') == 'New York'
assert closest_city('Toronto') == 'New York'
assert closest_city('Hong Kong') == 'Paris'
```
