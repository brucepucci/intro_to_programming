#### **Text cells**

Text cells are non-executable and are usually used to document or communicate some piece of information. The cell this is written in is a text cell. You can format text using the "Markdown" syntax. Information on Markdown can be found [here](https://github.com/adam-p/markdown-here/wiki/Markdown-Cheatsheet). Double clicking on a text cell will allow you to edit it.
