```python
import string
exclude = set(string.punctuation)
speech = ''.join(ch for ch in speech if ch not in exclude)
```

```python
speech = speech.upper()
```

```python
speech_list = speech.split(' ')
```

```python
word_counts = dict()
for word in speech_list:
    if word not in word_counts:
        word_counts[word] = 1
    else:
        word_counts[word] += 1
```

```python
sorted(word_counts.items(), key=lambda x: -x[1])
```
